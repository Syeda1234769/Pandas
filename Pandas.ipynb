# COMBINING DATAFRAMES

```python
# 1.concat()
df1 = pd.DataFrame({
    "Name": ["Azka", "Fatima","Marium"],
    "Marks": [80, 90, 60]
})

df2 = pd.DataFrame({
    "Name": ["Ramsha", "Wajeeha","Maria"],
    "Marks": [75, 85, 65]
})
final_df = pd.concat([df1, df2], ignore_index = True)
final_df
```

```python
# 2.merge()
students = pd.DataFrame({
    "ID": [19, 16, 21],
    "Name": ["Azka", "Fatima","Marium"]
})

marks = pd.DataFrame({
    "ID": [16, 21, 19],
    "Marks": [80, 90, 75]
})
final_df = pd.merge(students, marks, on="ID")

print(final_df)
```
